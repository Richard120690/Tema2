```python
import sqlite3
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from math import sqrt
from sklearn import preprocessing
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
import numpy as np
```

Creación de la conexión a la base de datos para leer la tabla "Player Attributes"

```python
cnx = sqlite3.connect('database.sqlite')
```

```python
df= pd.read_sql_query("SELECT* FROM Player_Attributes",cnx)
df.head(100)
```

### De todas las columnas, determinaremos cuales se usaran como "features o caracteristicas", para alimentar al modelo

```python
features= ['potential','crossing', 'finishing', 'heading_accuracy',
       'short_passing', 'volleys', 'dribbling', 'curve', 'free_kick_accuracy',
       'long_passing', 'ball_control', 'acceleration', 'sprint_speed',
       'agility', 'reactions', 'balance', 'shot_power', 'jumping',
       'strength', 'long_shots', 'aggression', 'interceptions',
       'vision', 'penalties', 'marking', 'standing_tackle', 'sliding_tackle',
       'gk_diving', 'gk_handling', 'gk_kicking', 'gk_positioning',
       'gk_reflexes']
```

### Seleccionamos el target

```python
target=['overall_rating']
```

### Limpiar los datos eliminando filas con valores nan

```python
df = df.dropna()
```

### Separamos las features en X, el target en Y

```python
x = df[features]
y = df[target]
```

### Separación de los datos en Trainign y Test Datasets

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.33, random_state=324)
```

Sesgo cuando una caracteristica o varias caracteristica puedan influir de más en el entrenamiento
### Utilizar un escalado estándar
Para que los datos de cada feature se acople a un rango y se eviten los sesgos de un feature con valores muy altos
que puedan influir de más en el target

```python
scale = preprocessing.StandardScaler()
```

### Entrenar con los datos escalados

```python
scale.fit(x_train)
```

```python
x_train = scale.transform(x_train)
```

```python
# Imprimir los nuevos valores minimo y máximo de las caracteristicas de entrenamiento
x_train.min()
```

```python
x_train.max()
```

### Creamos una instancia del modelo de Regresión lineal de Scikit Learn

```python
regresor = LinearRegression()
```

### Ajustando el modelo a los datos de entrenamiento
El ajuste de los datos del modelos del entrenamiento siempre se hace con los datos de entrenamiento

```python
regresor.fit(x_train, y_train)
```

### Realizar una prediccón usando el conjunto de datos de prueba (test set)

```python
# escalar los valores de prueba para que coincidan con los de entrenamiento
x_test = scale.transform(x_test)
```

```python
# hacer una predicción de el valor de salida, pasandole los valores de prueba
y_prediccion = regresor.predict(x_test)
```

```python
# obtener la diferencia que hay entre los valores de salida predichos contra los valores de salida reales
y_result = y_prediccion - y_test
y_result
```

### Calcular la raíz cuadrada del Error Caudrático Medio

```python
RMSE = sqrt(mean_squared_error (y_true = y_test, y_pred = y_prediccion))
print(RMSE)
```

### Mostrar la precisión que obtuvo el modelo al predecir con los valores de prueba

```python
regresor.score(x_test,y_test)
```
